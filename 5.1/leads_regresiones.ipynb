```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= pd.read_csv("leads_reales_clean.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
# Cambiamos los meses por números
df['Mes']=df['Mes'].replace({"Enero":1,
                                 'Febrero':2,
                                 "Marzo":3,
                                 "Abril":4,
                                 "Mayo":5,
                                 "Junio":6,
                                 "Julio":7,
                                 "Agosto":8,
                                 "Septiembre":9,
                                 "Octubre":10,
                                 "Noviembre":11,
                                 "Diciembre":12
                                 })
```

```python
df.head()
```

```python
df.info()
```

```python
# convertimos la variable "Mes" es tipo numérico
df['Mes'] = df['Mes'].astype(int)
```

```python
df.info()
```

```python
df= df.drop(['Año', "Mes"], axis=1)
```

```python
sns.pairplot(df)
```

```python
#Encontramos todas las correlaciones entre las variables 
Corr_Factors=df.corr()
Corr_Factors
```

```python
#Encontramos el valor absoluto de todas las correlaciones entre las variables 
Corr_Factors1= abs(Corr_Factors)
Corr_Factors1
```

```python
Heat_Map= sns.heatmap(Corr_Factors1, cmap = 'Blues', annot=True, fmt=".2f")  
Heat_Map
```

```python
#Imprimimos el scatter plot entre la variable dependiente e independiente 
#para observar el comportamiento en su dispersión 
from turtle import color
sns.scatterplot(x='Efectivos', y='Total', color="blue", data=df) 
#sns.scatterplot(x='speeding', y='total', color="red", data=df)
```

```python
Vars_Indep= df[['Efectivos']] ##en dos pares porque puede ser 1 o más
Var_Dep= df['Total'] #un solo par, es el único valor
```

```python
#Se define model como la función de regresión lineal
from sklearn.linear_model import LinearRegression
model= LinearRegression()
```

```python
model.fit(X=Vars_Indep, y=Var_Dep)
```

```python
#Verificamos los coeficientes obtenidos para el modelo ajustado
model.__dict__
```

```python
#Evaluamos la eficiencia del modelo obtenido por medio del coeficiente R Determinación
model.score(Vars_Indep,Var_Dep)
```

```python
y_pred = model.predict(X=df[['Efectivos']])
y_pred
```

```python
df.insert(0, 'Predicciones', y_pred)
```

```python
sns.scatterplot(x='Efectivos', y='Total', color="blue", data=df)
sns.scatterplot(x='Efectivos', y='Predicciones', color="red", data=df)
sns.lineplot(x='Efectivos', y='Predicciones', color="red", data=df)
```

```python
coef_Deter = model.score(X=Vars_Indep, y=Var_Dep)
coef_Deter
```

```python
coef_Correl = np.sqrt(coef_Deter)
coef_Correl
```
